###  DESCRIBE TABLE 

Muestra el schema basico:columnas, tipo de dato y nulabilidad. es equivalente a ver la estructura logica de la tabla. Util para revisar rapidamente campos antes de hacer un SELECT

```sql
DESCRIBE bronze_dev.bakehouse.customer_brz
```

### DESCRIBE EXTENDED

Incluye el schema y metadata adicional. Muestra propiedades de tabla y comentarios. Mas detallado que Describe Table

```sql
DESCRIBE EXTENDED bronze_dev.bakehouse.customer_brz
```

### DESCRIBE DETAIL

Devuelve informacion tecnicqa en formato tabular, incluye ubicacion fisica, tamañoi en bytes, nuemro de archivos. formato (delta), etc. Muy util para anlisis operativo y optimizacion

```sql
DESCRIBE DETAIL bronze_dev.bakehouse.customer_brz
```

### DESCRIBE HISTORY 

 Muestra el historial de versiones (time travel). incluye operacion(WRITE, MARGE, DELETE), usuario y timestamps. Clave para auditoria y debugging en pipelines

```sql
DESCRIBE HISTORY bronze_dev.bakehouse.customer_brz
```
