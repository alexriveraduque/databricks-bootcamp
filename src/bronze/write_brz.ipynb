```python
from pyspark.sql.functions import current_timestamp,date_format
```

```python
#CUSTOMER
dbutils.widgets.text("source_table_customer","samples.bakehouse.sales_customers") # customer
dbutils.widgets.text("target_table_customer","bronze_dev.bakehouse.customer_brz") # customer

#FRANCHISES
dbutils.widgets.text("source_table_franchises","samples.bakehouse.sales_franchises") # customer
dbutils.widgets.text("target_table_franchises","bronze_dev.bakehouse.franchises_brz") # customer
```

```python
source_customer = dbutils.widgets.get("source_table_customer")
target_customer = dbutils.widgets.get("target_table_customer")

source_franchises = dbutils.widgets.get("source_table_franchises")
target_franchises = dbutils.widgets.get("target_table_franchises")
```

```python
df_customer = spark.read.table(source_customer)
df_franchises = spark.read.table(source_franchises)
```

```python
df_final_customer = df_customer.select(
    df_customer.customerID.cast("String").alias("customerID"),
    df_customer.first_name.cast("String").alias("first_name"),
    df_customer.last_name.cast("String").alias("last_name"),
    df_customer.email_address.cast("String").alias("email_address"),
    df_customer.phone_number.cast("String").alias("phone_number"),
    df_customer.address.cast("String").alias("address"),
    df_customer.city.cast("String").alias("city"),
    df_customer.state.cast("String").alias("state"),
    df_customer.country.cast("String").alias("country"),
    df_customer.continent.cast("String").alias("continent"),
    df_customer.postal_zip_code.cast("String").alias("postal_zip_code"),
    df_customer.gender.cast("String").alias("gender"),
    current_timestamp().alias("ingest_timestamp"),
    current_timestamp().cast("date").alias("ingest_date"),
)

df_final_franchises = df_franchises.select(
    df_franchises.franchiseID.cast("String").alias("franchiseID"),
    df_franchises.name.cast("String").alias("name"),
    df_franchises.city.cast("String").alias("city"),
    df_franchises.district.cast("String").alias("district"),
    df_franchises.zipcode.cast("String").alias("zipcode"),
    df_franchises.country.cast("String").alias("country"),
    df_franchises.size.cast("String").alias("size"),
    df_franchises.longitude.cast("String").alias("longitude"),
    df_franchises.latitude.cast("String").alias("latitude"),
    df_franchises.supplierID.cast("String").alias("supplierID"),
    current_timestamp().alias("ingest_timestamp"),
    current_timestamp().cast("date").alias("ingest_date") 
)
```

```python
df_final_customer.write.mode("overwrite").saveAsTable(target_customer)
df_final_franchises.write.mode("overwrite").saveAsTable(target_franchises)
```
